#Выделение границ


```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
img = np.array(Image.open(r"/content/drive/MyDrive/3sem/MIOI/img3.jpg"))
# img = np.array(img)
plt.imshow(img)
print(img.shape)
```

```python
# convert to binary image
img = np.mean(img, axis=-1)
im = np.zeros(img.shape)

for i in range(img.shape[0]):
  for j in range(img.shape[1]):
    if img[i,j] > 127:
      im[i,j] = 1 
       
    else :
      im[i,j] = 0
```

```python
# function erosion

def er(image, mask):
    
    (W, H) = image.shape
    
    (f, f) = mask.shape
    pad = int((f - 1)/2)

    out_image = np.ones((W, H))

    #performing convolutional operation
    gray_image_array_pad = np.pad(image, ((pad,pad), (pad,pad)), mode='constant', constant_values = (0,0))
    for w in range(W):  
      for h in range(H):

        gray_image_array_slice= gray_image_array_pad[w:w+f, h:h+f]
        mul = np.multiply(mask , gray_image_array_slice)


        out_image[w, h] = np.max(mul)
        

    return out_image
```

```python
%matplotlib inline
```

```python
#structure element or mask
mask = np.ones((3,3))
mask2 = np.ones((21,21))
mask3 = np.ones((51,51))
```

```python
#erosion operation
er_im = er(im, mask)
kk = im-er_im
images = [im, er_im, kk]
images_names = ['Original', 'Erosion res','Result']
plt.figure(figsize=(25,25))
for id,ima in enumerate(images):
  plt.subplot(1,3,id+1)
  plt.title(images_names[id])
  plt.imshow(ima, cmap='gray')
```
