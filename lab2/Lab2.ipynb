#Оператор Собеля


```python
from PIL import Image
import matplotlib.pyplot as plt
img = Image.open("/content/drive/MyDrive/3sem/MIOI/4.png")
plt.imshow(img)
```

```python
import numpy as np
img_gray = np.mean(img, axis=-1)
plt.imshow(img_gray, cmap='gray')
```

```python
#ядра Собеля
sobel_matrix_1 = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])
sobel_matrix_2 = np.array([[1, 2, 1], [0, 0, 0], [-1, -2, -1]])

sob_image = np.zeros(img_gray.shape)
for i in range(1, len(sob_image)-1):
  for j in range(1, len(sob_image[0])-1):
    sob_image[i, j] = ((np.sum(img_gray[i-1:i+2, j-1:j+2] * sobel_matrix_1))**2 +
                       (np.sum(img_gray[i-1:i+2, j-1:j+2] * sobel_matrix_2))**2)**(1/2)

plt.imshow(sob_image, cmap='gray')
```
